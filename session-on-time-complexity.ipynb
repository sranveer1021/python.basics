```python
L = [ 1,2,3,4,5]

for i in range(0,len(L)//2):
    other = len(L) - i -1
    temp = L[i]
    L[i] = L[other]
    L[other] = temp

print(L)
```

```python
def fib(n):
    if n == 1 or n == 0:
        return 1
    else:
        return fib(n-1) + fib(n-2)
```
