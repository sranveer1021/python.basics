Q3:- Take 2 numbers as input from the user.Write a program to swap the numbers without using any special python syntax.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

temp = a
a = b
b = temp

print("After swapping:")
print("a =", a)
print("b =", b)
```

```python
### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.
```

```python
p1x = int(input('enter x cood of 1st point'))
p1y = int(input('enter y cood of 1st point'))
p2x = int(input('enter x cood of 2nd point'))
p2y = int(input('enter y cood of 2nd point'))

distance = ((p2x - p1x)**2 + (p2y - p1y)**2)**0.5

print(round(distance,2))
```

```python
### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

# For example:
# Input:
# heads -> 4
# legs -> 12
# <br>
# Output:
# dogs -> 2
# chicken -> 2



```

```python
heads = int(input("Enter total heads: "))
legs = int(input("Enter total legs: "))

dogs = (legs - 2 * heads) // 2
chicken = heads - dogs

print("dogs ->", dogs)
print("chicken ->", chicken)
```

```python
### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.
```

```python
a = int(input("Enter first term: "))
b = int(input("Enter second term: "))
n = int(input("Enter value of n: "))

d = b - a
nth_term = a + (n - 1) * d

print("Nth term of the AP:", nth_term)
```

```python
### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.
```

```python
n1 = int(input('num1'))
d1 = int(input('den1'))
n2 = int(input('num2'))
d2 = int(input('den2'))

rn = n1*d2 + n2*d1
rd = d1*d2

print('{}/{}'.format(rn,rd))
```

```python
### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



# Input:<br>
# Dimensions of the milk tank<br>
# H = 20cm, L = 20cm, B = 20cm
# <br><br>
# Dimensions of the glass<br>
# h = 3cm, r = 1cm
```

```python
import math

h_t = float(input('height'))
b_t = float(input('breadth'))
l_t = float(input('length'))

h_g = float(input('height of glass'))
r_g = float(input('radius of the glass'))

vol_tank = h_t*b_t*l_t
vol_glass = 3.14*r_g*r_g*h_g

print('no of glasses',math.floor(vol_tank/vol_glass))
```
