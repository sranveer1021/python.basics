# Tuples

A tuple in python is similar to a list. The diffrence between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

    In short, a tuple is an immutable list. A tuple can not be changed in any way once it is created.

Characteristics
- Ordered
- Unchangeble
- Allows duplicate


Plan of attack

 - Creating a Tuple
 - Accessing items
 - Editing items
 - Adding items
 - Deleting items
 - Operating on Tuples
 - Tuple Function

# Creating tuple

```python
t1 = ()
print(t1)
# create a tuple with a single item
t2 = ('hello',)
print(t2)
print(type(t2))
# homo
t3 = (1,2,3,4)
print(t3)
# hetro
t4 = (1,2,5,True,[1,2,3])
print(t4)
# tuple
t5 = (1,2,3,(4,5))
print(t5)
# using type conversion
t6 = tuple('hello')
print(t6)
```

# Diffrence between Lists and Tuples

- Syntax
- Mutability
- Speed
- Memory
- Build in functionality
- Error prone
- Usability

```python
#Speed
import time

L = list(range(100000000))
T = tuple(range(100000000))

start = time.time()
for i in L:
    i*5
print('List time',time.time()-start)

start = time.time()
for i in T:
    i*5
print('Tuple time',time.time()-start)
```

```python
# memory
import sys

L = list(range(1000))
T = tuple(range(1000))

print('List size',sys.getsizeof(L))
print('Tuple size',sys.getsizeof(T))
```

```python
a = [1,2,3]
b = a

a.append(4)
print(a)
print(b)
```

```python
a = (1,2,3)
b = a 

a = a + (4,)
print(a)
print(b)
```

# Sets

A set is an unordered collection of items.Every set element is unique(no duplicates) and must be immutable (cannot be changed).

However,a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations the union,intersection,symmetric difference,etc.

    Characterstics:
- Unordered
- Mutable
- No Duplicates
- Can't contain mutable data types


# Creating Sets

```python
# empty
s = set()
print(s)
print(type(s))
# 1D and 2D
s1 = {1,2,3}
print(s1)
#s2 = {1,2,3,{4,5}}
#print(s2)
# homo and hetro
s3 = {1,'hello',4,5,(1,2,3)}
print(s3)
# using type conversion
s4 = set([1,2,3])
print(s4)
# duplicates not allowed
s5 = {1,1,2,2,3,3}
print(s5)
# set can't have mutable items
s6 = {1,2,[3,4]}
print(s6)
```

```python
s1 = {1,2,3}
s2 = {3,2,1}

print(s1 == s2)
```

# Set operation

```python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1 | s2
# union(|)
# Intersection(&)
s1 & s2
# Difference(-)
s1 - s2
s2 - s1
# Symmetric Difference(^)
s1 ^ s2
# Membership Test
1 not in s1
# Iteration
for i in s1:
    print(i)
```

# Set Functions

```python
# len/sum/min/max/sorted
s = {3,1,4,5,2,7}
len(s)
sum(s)
min(s)
max(s)
sorted(s,reverse=True)
```

```python
# union/update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

#s1 | s2
s1.union(s1)

s1.update(s2)
print(s1)
print(s2)
```

```python
# intersection/intersection_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.intersection(s2)

s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
# difference/difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.difference(s2)

s1.difference_update(s2)
print(s1)
print(s2)
```

```python
# symmetric_difference/symmetric_difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.symmetric_difference(s2)

s1.symmetric_difference_update(s2)
print(s1)
print(s2)
```

```python
# isdisjoint/issubject/issuperset
s1 = {1,2,3,4}
s2 = {7,8,5,6}
s1.isdisjoint(s2)
```

```python
s1 = {1,2,3,45}
s2 = {7,8,5,6}

s1.issuperset(s2)
```

```python
# copy
s1 = {1,2,3}
s2 = s1.copy()

print(s1)
print(s2)
```

# Dictionary

Dictionary in Python is a collection of Keys values,used to store data values like a map.which,unlike other data types which hold onlt a single value as an element.

In some languages it is known as map or assosiative arrays.

dict = {'name':"Ranveer':,'age':23,'gender':'male'}

Characterstics:

 - Mutable
 - Indexing has no meaning
 - keys can't be duplicated
 - keys can't be mutable items

# Create Dictionary

```python
# empty dictionary
d = {}
d
# 1D dictionary
d1 = {'name':'ranveer','gender':'male'}
d1
# with mixed keys
d2 = {(1,2,3):1,'hello':'world'}
d2
# 2D dictionary ->JSON
s = {
    'name':'ranveer',
    'college':'bit',
    'sem':4,
    'subjects':{
        'dsa':50,
        'maths':54,
        'english':43
    }
}
s
# using sequence and dict fumction
d4 = dict([('name','nitish'),(2,2),(3,3)])
d4
# duplicate keys
d5 = {'name':'nitish','name':'rahul'}
d5
# mutable items as keys
d6 = {'name':'nitish',(123):2}
d6
```

# Dictionary Comprehension

{ key: value for vars in iterable }

```python
# print 1st 10 numbers and their squares
{i:i**2 for i in range(1,11)}
```

```python
# using existing dict
distances = {'delhi':1000,'mumbai':2000,'bangalore':3000}
{key:value*0.62 for (key,value) in distances.items()}
```

```python
# using zip
days = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Saturday"]
temp_C = [30.5,32.6,31.8,33.4,29.8,30.2,29.9]

{i:j for (i,j) in zip(days,temp_C)}
```

```python
# using if condition
products = {'phone':10,'laptop':0,'charger':32,'tablet':0}


{key:value for (key,value)in products.items() if value>0}
```

```python
# Nested Comprehension
# print tablets of number from 2 to 4
{i:{j:i*j for j in range(1,11)}for i in range(2,5)}
```
