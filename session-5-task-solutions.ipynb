# Tuple

Q1: Join Tuples if similar initial element

while working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element.
This problem can have applications in data domains such as Data Science

```python
Input : test_list = [(5,6), (5,7), (5,8), (6,10), (7,13)]
Output : [(5,6,7,8), (6,10), (7,13)]
```

```python
# write your code here
test_list = [(5,6), (6,8), (6,10), (6,7), (7,13), (5,7)]

unique = []

for i in test_list:
    unique.append(i[0])
unique = set(unique)

result = []
for i in unique:
    result.append([i])
    for j in test_list:
        if j[0] == i:
           result[-1].append(j[1])
            
print(list(map(tuple,result)))
```

# Q2: Multiply Adjacent elements(both side) and take sum of right and lest side multiplication result.

for eg.

```python
The original tuple : (1,5,7,8,10)
Resultant tuple after multiplication :

(1*5, 1*5+5*7, 7*8, 8*7 + 8*10, 10*8) -> (5,40,91,136,80)

Output-(3,40,91,136,80)
```

```python
# write your code here 
t = (1,5,7,8,10)

L = []

L.append(t[0]*t[1])

for i in range(1,len(t)-1):
    L.append(t[i]*t[i-1] + t[i]*t[i+1])

L.append(t[-1]*t[-2])
print(tuple(L))
```

# Q3: Check is tuples are same or not?

Two tuples would be same if both tuples have same element at same index

```python
t1 = (1,2,3,0)
t2 = (0,1,2,3)

t1 and t2 are not same
```

```python
# write your code here
t1 = (1,2,3,0)
t2 = (1,2,3,0)

flag = True
for i,j in zip(t1,t2):
    if i == j:
        continue
    else:
        flag = False
        break
    if flag:
        print('same')
    else:
        print('not same')
```

# Q4: Count no of tuples,list and set from a list


```python
list - [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi','bye'],['a','b']]
```

```python
Output:
```

```python
List-2
Sets-2
Tuple-1
```

```python
# write your code here
L = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi','bye'],['a','b']]
output = [0,0,0]

for i in L:
    if type(i) == list:
        output[0] = output[0] + 1
    elif type(i) == set:
        output[1] = output[1] + 1
    elif type(i) == tuple:
        output[2] = output[2] + 1
    else:
        pass

print('Lists.{}\nsets.{}\nTuples.{}'.format(output[0],output[1],output[2]))
```

### Q1: Write a program to find set of common elements in three lists using sets.

```python
Input : ar1 = [1,5,10,20,40,80]
        ar2 = [6,7,20,80,100]
        ar3 = [3,4,15,20,30,70,80,120]

output : [80, 20]
```

```python
# write your code here
ar1 = [1,5,10,20,40,80]
ar2 = [6,7,20,80,100]
ar3 = [3,4,15,20,30,70,80,120]

s1 = set(ar1)
s2 = set(ar2)
s3 = set(ar3)

result = list((s1 & s2) & s3)
print(result)
```
