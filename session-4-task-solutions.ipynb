# Problem 1: Combine two lists index-wise(columns wise)

Write a program to add two lists index-wise. Create a new list that contains the 0th index item from both the list, then the 1st index item,and so on till the least element. any leftover items will get added at the end of the new list.

Given List:

```python
list1 = ["M", "na", "i", "kh"]
list2 = ["y", "me", "s", "an"]
```

Output:

```python
[['M', 'y'], ['na', 'me'], ['i', 's'], ['kh', 'an']]
```

```python
# Write code here
list1 = ["M", "na", "i", "kh"]
list2 = ["y", "me", "s", "an"]

[[i,j] for (i,j) in zip(list1,list2)]
```

Problem 6: Find list of common unique items from two list. and show in increasing order

```python
num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]
```

```python
Output:
```

```python
[34,67,89]
```

```python
# Write code here
num1 = [23,45,67,78,89,34,67]
num2 = [34,89,55,56,39,67,67]

common = []

for i in num1:
    if i in num2:
        if i not in common:
            common.append(i)
        print(common)
```

Problem 7: Sort a list of alphanumric string based on product value of numeric character in it.if in any string ther is no numeric character take it's product value as 1.

 ['lac21','23fg','456','098d','1','kls']

Output:

```python
['lac21','23fg','lac21','1','kls','098d']
```

```python
# Write code hereL = ['lac21', '23fg', '456', '098d', '1', 'kls']

product_list = []

for item in L:
    product = 1
    has_digit = False
    for char in item:
        if char.isdigit():
            product *= int(char)
            has_digit = True

    # if no digit found, product should be 0
    if not has_digit:
        product = 0

    product_list.append(product)

# sort by product value (descending) and extract strings
result = [i[1] for i in sorted(zip(product_list, L), reverse=True)]

print(result)
```
