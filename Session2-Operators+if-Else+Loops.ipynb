## Operators in Python
- Arithmetric Operators
- Relational Operators
- Logical Operators
- Bitwise Operators
- Assignment Operators
- Membership Operators

```python
# Arithmetric Operators
print(5+6)

print(5-6)

print(5*6)

print(5/2)

print(5//2)

print(5%2)

print(5**2)
```

```python
# Relational Operators
print(4>5)

print(4<5)

print(4>=5)

print(4<=4)

print(4==4)

print(4!=4)
```

```python
# Logical Operators
print(1 and 0)

print(1 or 0)

print(not 1)
```

```python
# Bitwise Operators

# bitwise and
print(2 & 3)

# bitwise or
print(2| 3)

# bitwise xor
print(2 ^ 3)

print(~3)

print(4 >> 2)

print(5 << 2)
```

```python
# Assignment Operators

# =
# a = 2

a = 2

#a = a + 2
a +=2 
print(a)
```

```python
# Membership Operators

# in/not in

print('D' not in 'Delhi')

print(1 in [2,3,4,5,6])
```

```python
# menu driven calculation
fnum = int(input('enter the 1st number'))
snum = int(input('enter the 2nd number'))

op = input('Enter the operation')

if op == '+':
    print(fnum + snum)
elif op == '-':
    print(fnum - snum)
elif op == '*':
    print(fnum * snum)
else:
    print(fnum / snum)
```

```python
# While loop example -> program to print the table
# Program -> Sum of all digits of a given number
# Program -> Keep accepting numbers from users till he/she enters a ) and then find the avh
```

```python
number = int(input('enter the number'))

i = 1

while i<11:
    print(number,'*',i,'=',number * i)
    i += 1
```

```python
# while loop with else

x = 1

while x < 3:
    print(x)
    x +=1

else:
    print('limit crossed')
```

```python
# guessing game

# generate a random integer between 1 to 100
jackpot= random.randint(1,100)

guess = int(input('guess karo'))
counter = 1
while guess != jackpot:
    if guess < jackpot:
        print('galat!guess higher')
    else:
        print('galat!guess lower')
    guess = int(input('guess karo'))
    counter += 1

else:
    print('correct guess"')
    print('attemts',counter)
```
